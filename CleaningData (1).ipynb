## Step1: Load the Dataset

```python
import pandas as pd
```

```python
# Load the dataset

data = pd.read_csv("sample_data_cleaning_project.xlsx.csv")
print(data)
```

## step2: Ideentify Missing Values

```python
print("\nMissing Values count:")

print(data.isnull().sum())
```

## step3: Handle Missing Values

```python
data['Age']=data['Age'].fillna(data['Age'].median())
data['Salary']=data['Salary'].fillna(data['Salary'].mean())
data=data.dropna(subset=['Join_Date'])
print(data)
```

```python
## Step4:Check and remove duplicate rows
```

```python
data = data.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
data = data.apply(lambda x: x.str.lower() if x.dtype == "object" else x)
print(data)
```

```python
data = data.drop_duplicates(subset=['Name','Age'])
print(data)
```

```python
print(data.duplicated().sum())
```

## step5:Handale Outilers

```python
#calculate IQR for Salary
Q1 =  data['Salary'].quantile(0.25)
Q3 = data['Salary'].quantile(0.75)
IQR =Q3-Q1
#Filer out outilers
data =data[(data['Salary'] >= Q1-1.5 *IQR)& (data['Salary']<= Q3+1.5 *IQR)]
print(data)
```

## step6:Convert Data Type

```python
data['Join_Date'] =pd.to_datetime(data['Join_Date'])
print(data)
```

## step7:Encode Categorical variables

```python
#one-hot encode the 'Department' column
data = pd.get_dummies(data,columns=['Department'],drop_first=True)
#convert only the one-hot encode columns to integers
binary_columns = [col for col in data.columns if col.startswith('Department_')]
data[binary_columns]=data[binary_columns].astype(int)
print(data)
```

## step8: Save the dateset

```python
data.to_csv("cleaned_data.csv",index=False)
```

## Visualization

```python
import matplotlib.pyplot as plt
#Load my dataset 
data = pd.read_csv("cleaned_data.csv")
# Display the first few rows of the dataset
print(data.head())
```

## Bar Chart for Departments

```python
department_counts =data[['Department_HR','Department_IT']].sum()
#plotting department distribution
department_counts.plot(kind='bar',color=['blue','green'],figsize=(8,6))
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(ticks=[0,1],label=["HR","IT"],rotation=0)
plt.show()
           
           
```

```python
plt.figure(figsize=(8,6))
plt.hist(data['Salary'],bins=10, color='orange', edgecolor='black')
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

## Line plot for Age and Salary

```python
plt.figure(figsize=(8,6))
plt.plot(data['Age'], data['Salary'],marker='o', linestyle='-', color='purple')
plt.title("Salary vs. Age ")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

## Scatter plot for Salary and Department

```python
plt.figure(figsize=(8,6))
colors=['blue' if dept == 1 else 'green' for dept in data['Department_HR']]
plt.scatter(data['Salary'], data['Age'],c=colors, alpha=0.7,edgecolor='black')
plt.title("Salary vs. Age by Department( HR in Blue& IT in Green)")
plt.xlabel("Salary")
plt.ylabel("Age")
plt.grid(True)
plt.show()
```
